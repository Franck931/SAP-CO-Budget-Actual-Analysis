**Importation de Données**

```python
import pandas as pd
from IPython.display import display

df = pd.read_csv("sap_co_budget_reel_raw.csv")

display(df.head(10))
```

**Audit**

```python
# Valeurs manquantes

display(df.isnull().sum())
```

```python
ligne_manquantes = df[df.isnull().any(axis=1)]

display(ligne_manquantes)
```

```python
# Doublons
display(df.duplicated().sum())
```

```python
#Format de donnée
display(df["Mois"].dtype)
display(df["Mois"].unique())
```

```python
#Coherence des libelle

display(df["Business_Unit_Code"].unique())
display(df["Business_Unit"].unique())
display(df["Centre_Cout_Code"].unique())
display(df["Centre_Cout"].unique())
display(df["Budget_EUR"].unique())
display(df["Reel_EUR"].unique())
display(df["Nature_Depense"].unique())
```

```python
colonnes_avec_valeurs_espaces = []

for col in df.columns:
    if df[col].astype(str).str.contains(" ", na=False).any():
        colonnes_avec_valeurs_espaces.append(col)

display(colonnes_avec_valeurs_espaces)
```

```python
#Espace inutile

for col in ["Business_Unit", "Centre_Cout","Nature_Depense",]:
    print(f"\n--- {col} ---")
    
    valeurs_avec_espaces = df[col].astype(str)[
        df[col].astype(str).str.strip() != df[col].astype(str)
    ]
    
    display(valeurs_avec_espaces.unique())
```

```python
#Majuscules/minuscules
for col in ["Business_Unit", "Centre_Cout", "Nature_Depense"]:
    print(f"\n--- {col} ---")
    
    display(
        df[col]
        .astype(str)
        .str.strip()
        .str.lower()
        .value_counts()
    )
```

```python
#Montant suspect
display(df["Budget_EUR"].describe())
display(df["Reel_EUR"].describe())
```

```python
#plus gros montant(Budget_EUR)

display(df.nlargest(10, "Budget_EUR"))

#plus gros montant(Reel_EUR)
display(df.nlargest(10, "Reel_EUR"))
```

```python
#Valeur extreme(Budget_EUR)

Q1 = df["Budget_EUR"].quantile(0.25)
Q3 = df["Budget_EUR"].quantile(0.75)

IQR = Q3 - Q1

borne_basse = Q1 - 1.5 * IQR
borne_haute = Q3 + 1.5 * IQR

suspects_budget = df[
    (df["Budget_EUR"] < borne_basse) |
    (df["Budget_EUR"] > borne_haute)
]

display(suspects_budget)
```

```python
#Valeur extreme (reel_EUR)

Q1 = df["Reel_EUR"].quantile(0.25)
Q3 = df["Reel_EUR"].quantile(0.75)

IQR = Q3 - Q1

borne_basse = Q1 - 1.5 * IQR
borne_haute = Q3 + 1.5 * IQR

suspects_reel = df[
    (df["Reel_EUR"] < borne_basse) |
    (df["Reel_EUR"] > borne_haute)
]

display(suspects_reel)
```

```python
#Nombre de business unit

nombre_business_units = df["Business_Unit"].nunique()

display("Nombre de Business Units :", nombre_business_units)
```

**Nettoyage**

```python
#Valeur manquante

display(df[df["Budget_EUR"].isna()])
```

```python
#remplace la valeur manquante
df.loc[29,"Budget_EUR"] = 42500

display(df.loc[29])
```

```python
#Format de donnée
display(df["Mois"].unique())
```

```python
#uniformiser le format date
df["Mois"] = df["Mois"].astype(str).str.replace("/","-",regex=False)

df["Mois_date"] = pd.to_datetime(
    df["Mois"],
    format="%Y-%m",
    errors="coerce"
)

display(df["Mois_date"].dt.strftime("%Y-%m").unique())
```

```python
#Supprimer les espaces inutiles
colonnes_libelles = [
    "Business_Unit",
    "Centre_Cout",
    "Nature_Depense"
]

for col in colonnes_libelles:
    df[col] = df[col].astype("string").str.strip()
```

```python
for col in colonnes_libelles:
    print(f"\n{col}")
    display(df[col].unique())
```

```python
#Uniformiser les majuscules/minuscule
for col in colonnes_libelles:
    df[col] = df[col].str.strip().str.upper()
```

```python
for col in colonnes_libelles:
    print(f"\n--- {col} ---")
    display(df[col].value_counts())
```

```python
#exporter le fichier nettoyer
df.to_excel("sap_co_budget_reel_raw_cleaned.xlsx", index=False)

display("Fichier Excel créé avec succè !")
```
